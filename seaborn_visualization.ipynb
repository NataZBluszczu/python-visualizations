### Wykresy dystrybucji +

Podstawowe typy wykresów dystrybucji mogą być dodatkowo rozszerzone o więcej informacji. Wizualizacja zależności 2 zmiennych może przedstawiać na marginesach wizualizacje rozkładu każdej zmiennej osobno. Przy wybraniu większej liczby zmiennych, możliwe jest wygenerowanie wszystkich kombinacji pomiędzy nimi. Służą do tego funkcje `jointplot` i `pairplot`, które znajdują się na poziomie *figure*.

##### ⭐ Zadanie 8:

Przygotuj wykres punktowy z rozkładem gęstości na osiach marginalnych (`jointplot`) opierając się na danych, które sam wybierzesz w sensowny sposób. Uwzględnij co najmniej 3 serie danych. Zapoznaj się z obiektem `plot_joint` i narysuj dodatkowo rozkład gęstości wartości na głównym wykresie punktowym. Zadbaj o czytelność wykresu (tytuł wykresu, podpisy osi oraz legenda).

```python

genres = ['classical', 'pop', 'metal']
df_filtered = df[df['track_genre'].isin(genres)].copy()


df_sample = df_filtered.groupby('track_genre').sample(n=400, random_state=42)

sns.set_theme(style="white", palette="Set1")

g = sns.jointplot(
    data=df_sample,
    x='energy',
    y='loudness',
    hue='track_genre',
    alpha=0.5,
    s=15
)

g.plot_joint(sns.kdeplot, alpha=0.8, fill=False, linewidths=1.5, zorder=10, levels=4)


g.fig.suptitle('relacja energii i głośności z rozkładem gęstości konturowej',
               fontsize=16, y=1.05)


g.set_axis_labels('Energia (skala 0.0 - 1.0)', 'Głośność (w dB)', fontsize=12)

g.ax_joint.legend(title='Gatunek muzyczny')

g.ax_joint.grid(True, linestyle='--', alpha=0.4)

plt.show()
```

### Wykresy kategorialne

Wykresy kategorialne przedstawiają dane podzielone na kategorie. Funkcja `catplot` znajduje się na poziomie *figure* (poziom wyższy) i korzysta z funkcji z poziomu *axes* (poziom niższy): `stripplot`, `swarmplot`, `boxplot`, `violinplot`, `pointplot` oraz `barplot`. Do wizualizacji danych można wykorzystywać wszystkie powyższe funkcje z odpowiednimi parametrami.

##### ⭐ Zadanie 10:

Przygotuj wykres punktowy (`catplot`) opierając się na danych, które sam wybierzesz w sensowny sposób. Uwzględnij co najmniej 3 kategorie i co najmniej 2 serie danych. Zapoznaj się z dwiema metodami rysowania wykresów punktowych kategorialnych (`strip`, `swarm`) i zdecyduj, która będzie lepsza w twoim rozwiązaniu. Zapoznaj się z parametrem `order` i dobierz dla niego takie wartości, żeby na osi X etykiety były pokazane w kolejności innej niż domyślna. Zadbaj o czytelność wykresu (tytuł wykresu i podpisy osi).

```python

genres = ['jazz', 'metal', 'pop', 'hip-hop']
df_filtered = df[df['track_genre'].isin(genres)].copy()

df_sample = df_filtered.groupby('track_genre').sample(n=150, random_state=42)
df_sample['Wulgarne?'] = df_sample['explicit'].map({True: 'Tak', False: 'Nie'})

custom_order = ['metal', 'jazz', 'pop', 'hip-hop']
my_palette = ['#06d6a0', '#ef476f']

fig, axes = plt.subplots(1, 2, figsize=(16, 6), sharey=True)

sns.stripplot(
    data=df_sample,
    x='track_genre',
    y='danceability',
    hue='Wulgarne?',
    order=custom_order,
    jitter=0.25,
    alpha=0.7,
    palette=my_palette,
    ax=axes[0]
)

axes[0].set_title('Rozkład taneczności wg gatunku i wulgarności (strip)')
axes[0].set_xlabel('Gatunek muzyczny')
axes[0].set_ylabel('Taneczność (Danceability)')
axes[0].legend(title='Wulgarne treści')

sns.swarmplot(
    data=df_sample,
    x='track_genre',
    y='danceability',
    hue='Wulgarne?',
    order=custom_order,
    alpha=0.7,
    size=5,
    palette=my_palette,
    ax=axes[1]
)

axes[1].set_title('Rozkład taneczności wg gatunku i wulgarności (swarm)')
axes[1].set_xlabel('Gatunek muzyczny')
axes[1].set_ylabel('')
axes[1].legend(title='Wulgarne treści')

plt.suptitle('Porównanie metod prezentacji rozkładu taneczności', fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```

##### ⭐ Zadanie 11:

Przygotuj wykres pudełkowy (`catplot`) opierając się na danych, które sam wybierzesz w sensowny sposób. Uwzględnij co najmniej 3 kategorie oraz co najmniej 2 serie danych. Zapoznaj się z dwiema metodami rysowania wykresów pudełkowych kategorialnych (`box`, `boxen`) i zdecyduj, która będzie lepsza w twoim rozwiązaniu. Zadbaj o czytelność wykresu (tytuł wykresu i podpisy osi).

```python
genres = ['pop', 'hip-hop', 'rock', 'metal', 'jazz']
df_filtered = df[df['track_genre'].isin(genres)].copy()

df_filtered['Tonacja'] = df_filtered['mode'].map({1: 'Durowa (Wesoła)', 0: 'Mollowa (Smutna)'})

categories_order = ['pop', 'hip-hop', 'rock', 'metal', 'jazz']

fig, axes = plt.subplots(1, 2, figsize=(18, 7), sharey=True)

fig.suptitle('Rozkład popularności wg gatunku i tonacji (durowa vs mollowa)',
             fontsize=16, y=1.05)

palette = ['#f07167','#0081a7']

sns.boxenplot(
    data=df_filtered,
    x='track_genre',
    y='popularity',
    hue='Tonacja',
    ax=axes[0],
    palette=palette,
    linewidth=1,
    order=categories_order
)

axes[0].set_title('Metoda boxenplot (idealna dla dużych zbiorów)', fontsize=13, pad=10)
axes[0].set_xlabel("Gatunek muzyczny", fontsize=12)
axes[0].set_ylabel("Popularność (skala 0-100)", fontsize=12)
axes[0].tick_params(axis='x', labelsize=11)
axes[0].legend(title='Rodzaj tonacji', fontsize=10, loc='upper right')
axes[0].grid(axis='y', alpha=0.3, linestyle='--')

sns.boxplot(
    data=df_filtered,
    x='track_genre',
    y='popularity',
    hue='Tonacja',
    ax=axes[1],
    palette=palette,
    showfliers=True,
    order=categories_order
)

axes[1].set_title('Metoda boxplot (klasyczna)', fontsize=13, pad=10)
axes[1].set_xlabel("Gatunek muzyczny", fontsize=12)
axes[1].set_ylabel("") # Oś Y jest współdzielona
axes[1].tick_params(axis='x', labelsize=11)
axes[1].legend(title='Rodzaj tonacji', fontsize=10, loc='upper right')
axes[1].grid(axis='y', alpha=0.3, linestyle='--')

plt.tight_layout()
plt.show()
```

##### ⭐ Zadanie 12:

Przygotuj wykres kolumnowy (`catplot`) opierając się na danych, które sam wybierzesz w sensowny sposób. Uwzględnij co najmniej 3 kategorie oraz co najmniej 2 serie danych. Zapoznaj się z parametrem `palette` i dobierz dla niego nową wartość. Zadbaj o czytelność wykresu (tytuł wykresu i podpisy osi).

```python


genres = ['pop', 'hip-hop', 'rock', 'metal']
df_filtered = df[df['track_genre'].isin(genres)].copy()

df_filtered['Wulgarne treści'] = df_filtered['explicit'].map({True: 'Tak (Explicit)', False: 'Nie (Clean)'})


g = sns.catplot(
    data=df_filtered,
    x='track_genre',
    y='popularity',
    hue='Wulgarne treści',
    kind='bar',
    palette='mako',
    height=6.5,
    aspect=1.5,
    errorbar=None,          # Wyłączamy "czułki" błędu, by zrobić miejsce na napisy nad słupkami
    legend_out=False
)

g.figure.suptitle('Średnia popularność utworów wg gatunku i zawartości wulgaryzmów',
                  fontsize=16, y=1.04)
g.set_axis_labels('Gatunek muzyczny', 'Średnia popularność (skala 0-100)', fontsize=12)
g.tick_params(labelsize=11)

g.legend.set_title('Wulgarne treści')
g.legend.set_loc('upper right')

g.axes[0, 0].set_ylim(0, 80)

g.axes.flat[0].grid(axis='y', linestyle='--', alpha=0.5, color='gray')

for ax in g.axes.flat:
    for container in ax.containers:
        # fmt='%.1f' oznacza jedną liczbę po przecinku
        ax.bar_label(container, fmt='%.1f', fontsize=11, padding=5, weight='bold')

plt.tight_layout()
plt.show()
```

##### ⭐ Zadanie 13:

Przygotuj wykres kolumnowy (`catplot`) opierając się na danych, które sam wybierzesz w sensowny sposób. Uwzględnij co najmniej 3 kategorie oraz co najmniej 2 serie danych. Porównaj ze sobą cztery rodzaje słupków błędu: błąd standardowy (`se` - standard error), odchylenie standardowe (`sd` - standard deviation), przedział centylowy (`pi` - percentile interval) i przedział ufności (`ci` - confidence interval). Każdy rodzaj słupków błędu musi być przedstawiony na osobnym podwykresie (`subplot`) jednego obrazu. Zadbaj o czytelność wykresu (tytuł wykresu i podpisy osi).

```python

genres = ['pop', 'hip-hop', 'rock', 'jazz']
df_filtered = df[df['track_genre'].isin(genres)].copy()

df_filtered['Tonacja'] = df_filtered['mode'].map({1: 'Durowa (Wesoła)', 0: 'Mollowa (Smutna)'})

fig, axes = plt.subplots(2, 2, figsize=(14, 10), sharex=True, sharey=True)

error_types = ['se', 'sd', 'pi', 'ci']
titles = [
    'Błąd standardowy średniej (se)\n[Precyzja estymacji średniej]',
    'Odchylenie standardowe (sd)\n[Rozrzut wszystkich danych]',
    'Przedział centylowy (pi)\n[Gdzie leży 95% typowych utworów]',
    'Przedział ufności (ci)\n[Pewność co do prawdziwej średniej]'
]

for ax, err, title in zip(axes.flat, error_types, titles):
    sns.barplot(
        data=df_filtered,
        x='track_genre',
        y='danceability',
        hue='Tonacja',
        errorbar=err,         # ZMIENNY PARAMETR BŁĘDU
        palette='Set2',
        capsize=0.1,          # Dodaje poziome "daszki" do linii błędu
        ax=ax
    )

    ax.set_title(title, fontsize=12, fontweight='bold', pad=10)
    ax.set_xlabel('Gatunek muzyczny' if ax in axes[1, :] else '', fontsize=11)
    ax.set_ylabel('Taneczność (skala 0-1.0)' if ax in axes[:, 0] else '', fontsize=11)

    if ax != axes[0, 0]:
        ax.get_legend().remove()
    else:
        ax.legend(title='Tonacja', loc='lower right')

plt.suptitle('Porównanie metod wyznaczania słupków błędu w seaborn', fontsize=16, y=1.03)
plt.tight_layout()
plt.show()
```
